```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn.neural_network
import sklearn.metrics
from sklearn.metrics import classification_report,confusion_matrix

npz_file = np.load('FMNISTSmall.npz')
#print(npz_file.files)
train_img = npz_file['trainImgSmall']
#print("train_img.shape", train_img.shape)
train_lab = npz_file['trainLabelSmall']
#print("train_lab.shape", train_lab.shape)
test_img = npz_file['testImgSmall']
#print("test_img.shape", test_img.shape)
test_lab = npz_file['testLabelSmall']
#print("test_lab.shape", test_lab.shape)

# TO PLOT AN IMAGE:
#plt.figure()
#plt.imshow(train_img[2217,:].reshape(28,28), cmap='gray')
```

# Task 1: mlp regressor

Feed-forward neural network with 20 neurons in a single hidden layer and a single neuron in the output layer. 



```python
mlp_regressor_20 = sklearn.neural_network.MLPRegressor(20)
mlp_regressor_20.fit(train_img,train_lab)
predictions_20 = mlp_regressor_20.predict(test_img)
```

Function to interpret the output of the network in the following way: a value less than 0 is interpreted as zero, a value greater than 9 is considered to be 9 and a value between 0 and 9 is rounded to the nearest integer.

```python
for i in range(len(predictions_20)):
    if(predictions_20[i] < 0):
        predictions_20[i] = 0
    elif(predictions_20[i] > 9):
        predictions_20[i] = 9
    else:
        predictions_20[i] = round(predictions_20[i],0)  
```

Accuracy of this network:

```python
accuracy_20 = sklearn.metrics.accuracy_score(test_lab,predictions_20)*100
print(accuracy_20)
```

Finding the best network by increasing or decreasing the number of neurons within the (**single**) hidden layer. Saving this trained network to include it in the submission: 

```python
#gridsearch 
mlp_regressor = sklearn.neural_network.MLPRegressor()
#parameters = {'hidden_layer_sizes': range(1,101), 'solver': {‘lbfgs’, ‘sgd’, ‘adam’}, 'activation': {‘identity’, ‘logistic’, ‘tanh’, ‘relu’}, 'max_iter': range(100,1000,100)}
#grid_search = sklearn.model_selection.GridSearchCV(mlp_regressor,parameters)
#grid_search.fit(train_img,train_lab)
#print(grid_search)

#We've found out that the best network is the following
best_regressor = sklearn.neural_network.MLPRegressor(hidden_layer_sizes = 20, solver = "lbfgs", activation = "tanh", max_iter = 1000)
best_regressor.fit(train_img,train_lab)
predictions_br = best_regressor.predict(test_img)

for i in range(len(predictions_br)):
    if(predictions_br[i] < 0):
        predictions_br[i] = 0
    elif(predictions_br[i] > 9):
        predictions_br[i] = 9
    else:
        predictions_br[i] = int(round(predictions_br[i],0))

accuracy_br = sklearn.metrics.accuracy_score(test_lab,predictions_br)*100
print(accuracy_br)  
```

Ilustration of the results for the 20 hidden neuron network and the best network: 

```python
#20 hidden neurons network confusion matrix:
print(confusion_matrix(test_lab, predictions_20))
print(confusion_matrix(test_lab, predictions_br))
```
